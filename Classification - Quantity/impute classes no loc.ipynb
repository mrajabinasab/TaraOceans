```python
import pandas as pd
```

```python
df = pd.read_csv('no_loc_updated.csv')
df['depth, nominal']
```

```python
df['depth bathy']
```

```python
# Select columns with at least one NaN
nan_columns = df.columns[df.isna().any()]
df_nan = df[nan_columns]
df_nan
```

```python
df_nan['mhw_count'] =df['mhw_count']
sorted_df = df_nan.sort_values(by='mhw_count')
sorted_df
```

```python
class_0 = sorted_df[sorted_df['mhw_count'] == 0]
df_0 = pd.DataFrame(class_0.mean(), index = sorted_df.columns)
trans_0 = df_0.transpose()
trans_0
```

```python
class_1 = sorted_df[sorted_df['mhw_count'] == 1]
df_1 = pd.DataFrame(class_1.mean(), index = sorted_df.columns)
trans_1 = df_1.transpose()
```

```python
class_2 = sorted_df[sorted_df['mhw_count'].isin([2, 3])]
print(class_2)
df_2 = pd.DataFrame(class_2.mean(), index = sorted_df.columns)
trans_2 = df_2.transpose()
trans_2
```

```python
merged_df = trans_0.merge(trans_1, how = 'outer')
merged_df_1 = merged_df.merge(trans_2, how = 'outer')
merged_df_1
```

```python
fin_mean = pd.DataFrame(merged_df_1.mean(), index = sorted_df.columns)
trans_fin = fin_mean.transpose()
trans_fin
```

```python
# Fill NaN values in df_nan using trans_fin's first row where columns match
for column in df_nan.columns:
    if column in trans_fin.columns:
        fill_value = trans_fin[column].iloc[0]  # Use the first row's value
        df_nan[column] = df_nan[column].fillna(fill_value)
df_nan
```

```python
done = df_nan.to_csv('impute_mean_classes_no_loc.csv', index = False)
```
