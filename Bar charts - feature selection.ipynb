```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(3)
y1 = [94.44, 93.7, 96.46]
y2 = [93.7, 93.87, 96.43]
width = 0.2

# plot data in grouped manner of bar type
plt.bar(x-0.2, y1, width, color='cyan')
plt.bar(x, y2, width, color='orange')
plt.xticks(x, ['Variance', 'Correlation', 'Mutual information'])
plt.xlabel("Feature selection method")
plt.ylabel("Mean accuracy scores")
plt.legend(["With locations", "Without locations"])
plt.ylim(90, 100)
plt.show()
```

```python
fruits = ['Variance', 'Correlation', 'Mutual information']
sales = [93.6, 93.48, 96.65]

plt.barh(fruits, sales)
plt.xlabel("Mean accuracy scores in percentage")
plt.ylabel("Feature selection method")
plt.xlim(90, 100)
plt.show()
```
