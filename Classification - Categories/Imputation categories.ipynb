```python
import pandas as pd
from sklearn.utils import shuffle
from sklearn import preprocessing
```

```python
def numeric_columns(df):
    numeric_cols = []
    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            numeric_cols.append(col)
    return numeric_cols
# Load dataset
df = pd.read_csv("no_loc_updated.csv")

# Use function to extract numeric columns
numeric_cols = numeric_columns(df)
# print(df_num.shape)
# print(df_num.columns)
# print(df_num['MHW_count'].value_counts())

# Create a copy of the DataFrame
df_numeric = df[numeric_cols].copy()

# Now you can safely modify df_numeric
df_numeric['mhw_category'] = df['mhw_category']

replace_category = {
    'none': 0,
    'Moderate': 1,
    'Strong': 2,
    'Severe': 3
}

df_numeric['mhw_category'] = df['mhw_category'].replace(replace_category)

print(df_numeric['mhw_category'].value_counts())

# Shuffle the DataFrame
shuffled_df_numeric = shuffle(df_numeric, random_state=42)

# Separate features (X) and labels (y) after shuffling
X = shuffled_df_numeric.drop(['mhw_category', 'mhw_count'], axis=1)
y = shuffled_df_numeric['mhw_category']
X.shape, y.shape
```

```python
shuffled_df_numeric.dtypes!= 'float64'
```

```python
# Select columns with at least one NaN
nan_columns = shuffled_df_numeric.columns[shuffled_df_numeric.isna().any()]
df_nan = shuffled_df_numeric[nan_columns]
df_nan
```

```python
class_0 = shuffled_df_numeric[shuffled_df_numeric['mhw_category'] == 3]
df_0 = pd.DataFrame(class_0.mean(), index = shuffled_df_numeric.columns)
trans_0 = df_0.transpose()
trans_0
```

```python
class_1 = shuffled_df_numeric[shuffled_df_numeric['mhw_category'] == 0]
df_1 = pd.DataFrame(class_1.mean(), index = shuffled_df_numeric.columns)
trans_1 = df_1.transpose()
trans_1
```

```python
class_2 = shuffled_df_numeric[shuffled_df_numeric['mhw_category'] == 2]
df_2 = pd.DataFrame(class_2.mean(), index = shuffled_df_numeric.columns)
trans_2 = df_2.transpose()
trans_2
```

```python
class_3 = shuffled_df_numeric[shuffled_df_numeric['mhw_category'] == 1]
df_3 = pd.DataFrame(class_3.mean(), index = shuffled_df_numeric.columns)
trans_3 = df_3.transpose()
trans_3
```

```python
merged_df = trans_0.merge(trans_1, how = 'outer')
merged_df_1 = merged_df.merge(trans_2, how = 'outer')
merged_df_2 = merged_df_1.merge(trans_3, how = 'outer')
merged_df_2
```

```python
fin_mean = pd.DataFrame(merged_df_2.mean(), index = shuffled_df_numeric.columns)
trans_fin = fin_mean.transpose()
trans_fin
```

```python
# Fill NaN values in df_nan using trans_fin's first row where columns match
for column in df_numeric.columns:
    if column in trans_fin.columns:
        fill_value = trans_fin[column].iloc[0]  # Use the first row's value
        df_numeric[column] = df_numeric[column].fillna(fill_value)
df_numeric
```

```python
df_numeric.to_csv('Categories imputed.csv', index = False)
```
